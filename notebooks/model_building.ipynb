```python
#import requires libraries and modules to be used in the model building and evaluation process
import pandas as pd
import numpy as np
from sklearn import metrics
from sklearn.model_selection import train_test_split
from sklearn.metrics import(
    confusion_matrix,
    classification_report,
    accuracy_score
)
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from imblearn.combine import SMOTEENN
```

```python
df=pd.read_csv("telco_churn.csv")
df.head()
```

```python
df=df.drop('Unnamed: 0',axis=1)
```

```python
x=df.drop('Churn',axis=1)
x
```

```python
y=df['Churn']
y
```

##### Train test split

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

##### Decision Tree Classifier

```python
model_dt=DecisionTreeClassifier(criterion='gini',random_state=100,max_depth=6,min_samples_leaf=8)
```

```python
model_dt.fit(x_train,y_train)
```

```python
y_pred=model_dt.predict(x_test)
y_pred
```

```python
model_dt.score(x_test,y_test)
```

```python
print(classification_report(y_test,y_pred))
```

 The Decision Tree achieved an accuracy of approximately **78.9%** on the test dataset.
- The model performs better in identifying **non-churn customers (Class 0)**, with a recall of **89%** and an F1-score of **86%**.
- For **churn customers (Class 1)**, the model achieved a precision of **64%** and a recall of only **52%**.
- The relatively low recall for Class 1 indicates that the model is **missing a considerable number of actual churn customers**.
- The dataset is **imbalanced**, with more non-churn customers than churn customers.
- Therefore, accuracy alone is not sufficient to evaluate the model.
- **SMOTEENN** can be applied to address class imbalance and potentially improve the identification of churn customers.

```python
sm=SMOTEENN()
x_resampled,y_resampled=sm.fit_resample(x,y)
```

```python
x_rtrain,x_rtest,y_rtrain,y_rtest=train_test_split(x_resampled,y_resampled,test_size=0.2)
```

```python
model_dt_smote=DecisionTreeClassifier(criterion='gini',random_state=100,max_depth=6,min_samples_leaf=6)
```

```python
model_dt_smote.fit(x_rtrain,y_rtrain)
y_rpred=model_dt_smote.predict(x_rtest)
model_score_r=model_dt_smote.score(x_rtest,y_rtest)
print(model_score_r)
print(classification_report(y_rtest,y_rpred))
```

```python
print(metrics.confusion_matrix(y_rtest,y_rpred))
```

-Now we can see better results i.e., **accuracy 93%** and good recall, precision, f1-score
